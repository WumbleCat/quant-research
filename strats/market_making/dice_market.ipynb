```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# simulate random dice rolls
rolls = np.random.randint(1,7,10000)
```

```python
average_score = np.cumsum(rolls)/ np.arange(1, len(rolls) + 1)
```

```python
plt.plot(average_score)
plt.axhline(3.5, color='r')
```

```python
print(rolls.mean())
```

```python
plt.plot(rolls[:50])
plt.axhline(3.5, color='r', linestyle='--')
```

```python
bid = 3
ask = 4

portfolio = 0
profit = 100

portfolio_history = []
profit_history = []

portfolio_history.append(portfolio)
profit_history.append(profit)

for roll in rolls:
    if roll <= bid: # 3
        if profit >= 0:
            portfolio += 1
            profit -= roll

    if roll >= ask: # 4

        if portfolio >= 0:
            portfolio -= 1
            profit += roll

    portfolio_history.append(portfolio)
    profit_history.append(profit)

print(portfolio)
```

```python
plt.plot(portfolio_history)
```

```python
plt.plot(profit_history)
```
